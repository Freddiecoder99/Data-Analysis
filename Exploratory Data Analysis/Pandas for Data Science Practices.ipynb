```python
# Pandas: Effient Python Tool for Loading, Cleaning, Sorting and Filtering Structured Data(Tables-Spreadsheet, SQL)
# Two Data Structures:
# 1. Pandas Series: Single Column Data(One dimentional Data)
# 2. Pandas DataFrame: Multi-Column Data(Two Dimensition Data) -> rows and columns


# PANDAS SERIES 1D
# By Default: Each Data is assigned an index number starting from zero, Also the column name is an index
import pandas as pd
stock_prices = [180, 185, 190, 200, 195, 210, 220]
days =  ["Mon", "Tue", "Wed", "Thur", "Fri", "Sat", "Sun"]

stock_prices_series =  pd.Series(stock_prices, name = "Stock Prices", index = days)
stock_prices_series
```

```python
import pandas as pd

# Creating a Pandas Series
stock_prices = pd.Series([180, 185, 190, 200, 195, 210, 220],
                         index=["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"], name = "Stock Prices")

stock_prices
```

```python
# Properties of Pandas Series
# Series have 4 Properties
# 1. Values -> Information Represented in a Numpy Array
# 2. Index  -> How the information can be Retrieved, By Default index are numerical starting from zero
# 3. Name   -> Represent the name of the Column
# 4. Dtype  -> Datatypes of values stored in a Pandas Series

print(stock_prices.values)
print(type(stock_prices.values))
print(stock_prices.index)
print(stock_prices.name)
print(stock_prices.dtype)

```

```python
# Common Operations on Pandas Series
# 1. Statistical Operations
max_price = stock_prices.max()
print(max_price)

min_price = stock_prices.min()
print(min_price)

mean_price = stock_prices.mean()
print(mean_price)

median_price = stock_prices.median()
print(median_price)
```

```python
# 2. Filtering Operations: Extract Information
# a) Find Days When Stock Price Was Above 190
stock_prices[stock_prices > 190]
```

```python
# b) Apply mathematical Operations

adjusted_prices = stock_prices * 1.10  # Increase by 5%
adjusted_prices
```

```python
# PANDAS DATAFRAME 2D
# Table(rows, columns) for Representing Structured Data
# Multiple Column Data, where each column represent a Pandas Series

# Employee Salary Data

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 22, 28],
    "Job Role": ["Chef", "Waiter", "Manager", "Cashier"],
    "Salary": [50000, 30000, 70000, 40000]
}

df = pd.DataFrame(data)
df
```

```python
# Properties of a DataFrame
# 1. shape: The number of rows and columns
# 2. size : The total items stored in the array

print(df.shape)
print(df.size)
```

```python
# Importing DataFrames from Different Sources
# To read data from Different Sources we can use the following functions depending on the data
# 1. Excel -> pd.read_excel(link)
# 2. CSV   -> pd.read_csv(link)
# 3. JSON  -> pd.read_json(link)


data_url = 'http://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data'
columns = [
    'age', 'workclass', 'fnlwgt', 'education', 'education_num',
    'marital_status', 'occupation', 'relationship', 'ethnicity',
    'gender','capital_gain','capital_loss','hours_per_week',
    'country_of_origin','income'
]

# Loading an external data
data = pd.read_csv(data_url, names = columns)
data
```

```python
# Common Operations in Pandas DataFrame
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 22, 28],
    "Job Role": ["Chef", "Waiter", "Manager", "Cashier"],
    "Salary": [50000, 30000, 70000, 40000]
}

df = pd.DataFrame(data)
df
```

```python
# a) Sorting Employees by Salary
df.sort_values(by = "Salary", ascending=False)
```

```python
#  b) Filtering Employees with Salary Above 40,000
df['Salary'] > 40000

df[df['Salary'] > 40000]
```

```python
# c)  Calculating Average Salary
df['Salary'].mean()
```

```python
# d) Adding a new Column
# Adding a New Column for Bonus (10% of Salary)

df['Bonus'] = df['Salary'] * 0.20
df
```

```python
import pandas as pd

climate_data = pd.DataFrame({
    "Year": [2000, 2005, 2010, 2015, 2020],
    "Avg Temperature": [14.3, 14.5, 14.8, 15.1, 15.5]
})

# Calculate the temperature trend
climate_data['Temp Change'] = climate_data["Avg Temperature"].diff()
climate_data
```

```python
sales_data = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Monitor"],
    "Sales": [500, 1200, 300, 700],
    "Revenue": [500000, 900000, 150000, 350000]
})

sales_data

# Find the most profitable product
most_profitable = sales_data[sales_data["Revenue"] == sales_data["Revenue"].max()]
most_profitable
```

```python
hospital_data = pd.DataFrame({
    "Hospital": ["A", "B", "C", "D"],
    "Patients": [150, 200, 170, 220],
    "Doctors": [20, 30, 25, 35]
})

# Calculate doctor-to-patient ratio
hospital_data["Doctor-Patient Ratio"] = hospital_data["Patients"] / hospital_data["Doctors"]
hospital_data
```

```python
education_data = pd.DataFrame({
    "School": ["X", "Y", "Z", "W"],
    "Students": [500, 750, 600, 800],
    "Teachers": [40, 50, 45, 55]
})

# Calculate student-to-teacher ratio
education_data["Student-Teacher Ratio"] = education_data["Students"] / education_data["Teachers"]
education_data
```

```python
transport_data = pd.DataFrame({
    "City": ["New York", "London", "Tokyo", "Paris"],
    "Avg Commute Time (min)": [45, 40, 35, 50],
    "Public Transport Usage (%)": [70, 80, 85, 75]
})

# Find the city with the shortest commute time
best_commute = transport_data[transport_data["Avg Commute Time (min)"] == transport_data["Avg Commute Time (min)"].min()]
best_commute
```
