# Python Arithmetic Operators

Python provides several arithmetic operators for performing basic mathematical operations. Below are examples and explanations:


```python
print(3 + 2)   # addition(+)
print(3 - 2)   # subtraction(-)
print(3 * 2)   # multiplication(*)
print(3 / 2)   # division(/)
print(3 ** 2)  # exponential(**)
print(3 % 2)   # modulus(%)
print(3 // 2)  # Floor division operator(//)
```

# Understanding Data Types in Python

Data types in Python define the kind of data that can be stored and manipulated within a program. The `type()` function allows you to check the data type of any value or variable. Python provides a variety of built-in data types, including integers (`int`), which represent whole numbers, and floats (`float`), which represent decimal numbers. For complex numbers, Python uses the `complex` data type. Strings (`str`) are used for textual data, enclosed in single or double quotes. Lists (`list`) are mutable sequences used to store multiple items, while tuples (`tuple`) are immutable collections.

Dictionaries (`dict`) store data in key-value pairs, making them ideal for representing structured data. Sets (`set`) are unordered collections that automatically eliminate duplicates. Python also provides the boolean data type (`bool`) for logical operations, returning either `True` or `False`. For instance, `3 == 3` and `3 >= 3` both return `True`, demonstrating boolean logic. Understanding these data types is crucial for effective programming, as each type offers unique properties and methods that determine how data can be manipulated and processed.


```python
# Checking data types
# type(value)
print(type(10))                  # Int
print(type(3.14))                # Float
print(type(1 + 3j))              # Complex
print(type('Asabeneh'))          # String
print(type([1, 2, 3]))           # List
print(type({'name':'Asabeneh'})) # Dictionary - Map
print(type({'047': 'Nairobi', '001':'Mombasa', 'pin': 2343}))
print(type({9.8, 3.14, 2.7}))    # Set
print(type((9.8, 3.14, 2.7)))    # Tuple
print(type(3 == 3))              # Bool
print(type(3 >= 3))              # Bool
print(3==4)
print(2<3)
print(["Jane", "Mary"])
print(("Jane", "Mary"))
print({"Jane", "Mary"})

word = 'Python'
sentense = "I am enjoying Python, I will use it in Data Science"
multi_line = '''
I am enjoying Python.
I will use it in Data Science'''

print(word)
print(sentense)

```
